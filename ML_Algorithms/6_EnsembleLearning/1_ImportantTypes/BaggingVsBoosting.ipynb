# [Bagging vs Boosting](https://youtu.be/7M5oWXCpDEw?list=PLKnIA16_Rmvbr7zKYQuBfsVkjoLcJgxHH)  
### There are three main differences:  
#### 1. Types of model used:
   * **In Bagging**, base models that are used for training are Low Bias High Variance models... meaning they tend to overfit on their own individually(fully grow decision tree).
   * **In Boosting**, base models that used for training are High Bias and Low Variance models... meaning they would underfit if fitted on their own (decision stump: decision tree with max_Depth=1).

#### 2. Squential vs Parallel:  
* **In Bagging**, all the models are (/can be) trained parallely, it's possible because their training don't depend on each other. they all get different data to train on.
* **In Boosting**, all the models are trained in squential order... output of one model becomes the input of second model and they keep refinig their outcome model by model.

#### 3. Weightage of base models:  
* **In Bagging**, weightage of all models is same. meaning they get equal importance in the final prediction. all of their predictions are taken into consideration equally.
* **In Boosting**, weightage is not often equal, meaning some model's opinion matter more for final prediction then the others.
