### Three main concepts to know before starting AdaBoost.  
1. **Weak Learners**: Weak learners are the models with accuracies only a little above 50%.
2. **Decision Stumps**: Decision Stumps are the decision trees with max_depth=1 that are weak learners, meaning decision trees with accuracies a little above 50%. Its possible to use other models as weak learners in boosting as well but decision trees are most widely used and easy to monitor.
3. **+1 and -1**: if the problem is a binary classification we don't take 0 and 1 as classes in boosting, its -1 and +1.


--- 

**Stage wise aditive method**  This is a term used to describe adaboost, it adds weak models together to boost their performace stage wise.  


---

### General Idea of Working.  
![ytss](assets/1_ada_working.png)  

#### Steps taken during training process of AdaBoost:  
1. Take a data and run decision tree with depth = 1 on it... ofourse the one split has to be where information gain is maximum an dimpurity is minimum.
2. Next up, Do two things:
    * find its performance lets say by accuracy_score and then assign it a weightage constant lets say alpha based on how well its accuracy was, higher accuracy => higher weightage.
    * notice its mistakes and do upsampling on them... in general upsampling means giving the data points that miss-categorized more importance/focus. and send this data to next decision stump.
3. Next stump now focuses more on the upscaled data points and corrects the mistakes of previous stump by making the split where information is as maximum as possible and the upscalled data points are correctly categorized as well. But doing so it could also end up making its own mistakes... nevertheless, alpha is assigned to it as well based on its performance and miss-categorized data points are also upscaled here. <br>
   ...  <br>
   ...<br>
   ...<br>

These steps keep on repeating until we use all of our decision stumps... now we have all of our decisoin stumps trained.  
<br>

### How do we use all of these to actually make a prediction?  
The formula goes as :  
**`h(x) = sign(alpha1*h1(x) + alpha2*h2(x) + alpha3*h3(x) + ... +alphan*hn(x))`**  
*  every h from h1 to hn are our decision stumps.
*  x is the data being passed to it.
*  every alpha from 1 to n is the weightage constant of every model.
*  sign just tells us that either the result is +ve or -ve (meaning its the sign that matters not the value if purpose is just to make the prediction), if its positive it means that the new query is being classified as +1 and if its negative it means its being classified as -1 (this is also the reason why we don't take 1 and 0 as classes here in adaboost.).


So this is how we make the prediction.
