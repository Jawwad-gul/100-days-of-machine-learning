```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split,cross_val_score
import pandas as pd
import numpy as np
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
from sklearn.datasets import make_circles
import matplotlib.pyplot as plt
```

```python
ad1 = AdaBoostClassifier??
```

```python
X,y = make_circles(n_samples=500,factor=0.1,noise=0.35,random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
adb = AdaBoostClassifier()
np.mean(cross_val_score(adb,X,y,cv=10,scoring='accuracy'))
```

```python
adb.fit(X,y)
```

```python
def plot_decision_boundary(clf):
    x_range = np.linspace(X.min(),X.max(),100)
    xx1,xx2 = np.meshgrid(x_range,x_range)
    y_hat = clf.predict(np.c_[xx1.ravel(),xx2.ravel()])
    y_hat = y_hat.reshape(xx1.shape)
    plt.contour(xx1,xx2,y_hat,alpha=0.2)
    plt.scatter(X[:,0],X[:,1],c=y,cmap='viridis',alpha=0.9)
    plt.title('decision boundary')
    plt.show()
plot_decision_boundary(adb)
```

```python
abc = AdaBoostClassifier(n_estimators=1500,learning_rate=0.01)# learning rate reduces the increase of weights on every stage, which means smaller learning rates can help avoid overfitting. 
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
from sklearn.model_selection import GridSearchCV

params = {
    'learning_rate': [0.0001,0.001,0.01,0.1,1],
    'n_estimators':[10,50,100,500]
}
grid = GridSearchCV(AdaBoostClassifier(),param_grid=params,n_jobs=-1,cv=10,scoring='accuracy',verbose=5)
grid.fit(X,y)
```

```python
print(f"best score: {grid.best_score_}\n best params:  {grid.best_params_}")
```
