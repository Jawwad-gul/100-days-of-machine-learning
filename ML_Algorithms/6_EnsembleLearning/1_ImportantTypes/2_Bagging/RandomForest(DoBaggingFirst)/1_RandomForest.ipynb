### Decision Tree is Low Bias and High Variance Model, How does RnadomForest tackle that?  Simple:  
#### _(this is being revised same as it was in bagging)_
A model being Low Bias and High Variance means that it does pretty good on training data but is very sensetive to even the slight changes in training data... like a small change in training data will cause major leaps in its accuracy, simply put it isn't that much accurate on test data... **Overfits**.  

Now we know that DecisionTree is sensetive to change in training data due to its nature of being high Variance. Let's say we have data of 1000 samples and have trained DT, and now we replaced the 100 old samples with the new ones in the training data... now if there is only one model... it alone will have to take all the impact of those 100 new samples... its decision boundary will definetly make noticeable change. BUT what happens in case of RandomForest... since bootstrapping exists, all those new samples won't be just loaded onto single decisiontree model.. they will all be randomly divided into different models... one might get 10 out of 100 other might get 7, and so on... so the impact reduced to a very small number depending on the variation made and number of models and samples we are taking. Not just the fact that the impact has been reduced, at the end when aggregating... decision is made by generalizing all the predictions of base models by using majority vote method. Decision Boundary becomes even smoother. resulting in **Low Bias and Low Variance**  
<br>
  


## Bagging Vs Decision Tree. What's the difference?

What's the difference between Bagging with decisionTrees and RandomForest... the main difference is that when sampling is done in bagging... its tree level... meaning it won't change during the training process once the samples have been assigned to a tree they remain the same... WHILE in RandomForest each split of the tree occurs after resampling, so sampling is done on each node rather than just at the start, it can be called node level sampling whereas in bagging it was tree level sampling.
