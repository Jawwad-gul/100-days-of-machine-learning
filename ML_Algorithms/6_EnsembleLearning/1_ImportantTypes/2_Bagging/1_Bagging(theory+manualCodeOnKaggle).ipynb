**Bagging** word is made up from two words
* **Bootstrapping**
    Bootstrapping is a term related to statistics... it means extracting a sample data from original data for further processing. So in this case we extract k number of rows randomly for each model (**which by the way are always same in case of bagging**) separately from the original data.Upto the training step is called Bootstrapping.
* **Aggregation**.  
    Aggregation is the process of deciding the output... for example if we have a classification problem we would take majority's vote and return the final prediction, thats aggregation.

---


### Why Use Bagging?  
cuz we can reduce High variance to low variance and still have the same low bias. Commonly we don't get this situation,  
**Problem**
most of the models are either low bias, high variance or high bias, low variance (ofcourse tuning helps but in general its truth.) bagging helps in this regard to achieve low bias, low variance model How?  
So what we do is we select those models which are Low bias and High Variance as base models. meaning these models tend to fall to overfitting (individually). cuz of high variance. how do we know model is overfitted cuz of high variance? well if we change the certain number of rows to a different data from the original training data, if performance of model doesn't stay consistent instead drops noticeably, our model had high variance cuz it couldn't handle the slight variation in data.
Bagging simulates changes in data by itself to get rid of high varianceness. since bootstrapping draws out completely random data and distributes it to the models, the varied will affect each single model, but it won't be that big of an affect and since every model will take affect differently, the mistakes wil be cleared out during aggregation. so the greater affect reduces to small one. and ultimately final prediction can stay shielded from high variance's affect. this is how bagging tackles high variance and reduces it.
